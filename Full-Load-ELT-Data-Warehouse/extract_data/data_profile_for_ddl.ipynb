```python
all_dfs = {
    name : pd.read_csv(path)
    for name, path in all_data_files.items()
}
```

```python
def print_overview(entity_name) : 
    print(f"{'=' * 120}")
    print(f"TABLE: {entity_name.upper()}")
    print(f"{'=' * 120}")

    print(
        tabulate(
            all_dfs[entity_name].sample(10),
            headers='keys',
            tablefmt='simple_grid',
            showindex=False
        )
    )
```

```python
print_overview('aircraft')
print_overview('aircraft_maintenance_history') 
print_overview('aircraft_models')
print_overview('airlines')
print_overview('airports')
print_overview('ancillary_services')
print_overview('baggage')
print_overview('boarding_passes')
print_overview('bookings')
print_overview('check_in_events')
print_overview('crew_assignments')
print_overview('crew_members')
print_overview('customers')
print_overview('customer_support_tickets')
print_overview('feedback')
print_overview('flight_cancellations')
print_overview('flight_delays')
print_overview('flight_routes')
print_overview('flight_schedules')
print_overview('flights')
print_overview('flight_status_logs')
print_overview('fuel_usage')
print_overview('gate_information')
print_overview('loyalty_members')
print_overview('loyalty_transactions')
print_overview('maintenance_logs')
print_overview('passengers')
print_overview('payments')
print_overview('refunds')
print_overview('seat_assignments')
print_overview('tickets')
print_overview('weather_conditions')
```
