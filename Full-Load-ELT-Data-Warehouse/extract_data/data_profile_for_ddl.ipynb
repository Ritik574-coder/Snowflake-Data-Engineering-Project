```python
all_dfs = {
    name : pd.read_csv(path)
    for name, path in all_data_files.items()
}
```

```python
def print_summary(entity_name) : 
    print(
        tabulate(
            all_dfs[entity_name].sample(10),
            headers='keys',
            tablefmt='simple_grid',
            showindex=False
        )
    )
```

```python
print_summary('aircraft')
```
