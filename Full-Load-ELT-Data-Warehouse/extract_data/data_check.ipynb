### **AIRCRAFT SUMMERY FOR HIGH LEVEL OVERVIEW**

```python
def create_summary(df: pd.DataFrame) -> pd.DataFrame :

    """
    Create a column-level profiling summary for a Pandas DataFrame.

    This function performs an initial structural and data-quality assessment
    of each column in the dataset. It provides information about data types,
    completeness, cardinality, duplicate values, and memory consumption.

    The generated summary is intended for the initial data-profiling stage
    of an ETL/ELT pipeline before applying transformations, validation rules,
    or data-quality checks.
    """

    summary = pd.DataFrame(
            {
                "Column_Name"     : df.columns,
                "Data_Type"       : df.apply(lambda x : x.dtypes),
                "NotNull_Count"   : df.apply(lambda x : x.notnull().sum()),
                "Null_Count"      : df.apply(lambda x : x.isnull().sum()),
                "Unique_Count"    : df.apply(lambda x : x.nunique()),
                "Duplicate_count" : df.apply(lambda x : x.duplicated().sum()),
                "Memory_Usage_KB" : df.apply(
                    lambda x: round(x.memory_usage(deep=True) / 1024, 2)
            )
            }
        )

    return summary
```

```python
all_summaries = {
    name: create_summary(df)
    for name, df in all_dfs.items()
}
```

```python
def print_summary(entity_name) : 
    print(f"{'=' * 120}")
    print(f"ENTITY : {entity_name.upper()} | COLUMN_COUNTS : {len(all_summaries[entity_name])} | ROW_COUNTS : {len(all_dfs[entity_name])}")
    print(f"{'=' * 120}")

    print(
        tabulate(
            all_summaries[entity_name],
            headers='keys',
            tablefmt='simple_grid',
            showindex=False
        )
    )
```

```python
from tabulate import tabulate

with open("data_profiling_summary.txt", "w", encoding="utf-8") as file:

    for entity_name, summary in all_summaries.items():

        file.write("=" * 100 + "\n")
        file.write(f"TABLE: {entity_name.upper()}\n")
        file.write("=" * 100 + "\n\n")

        file.write(
            tabulate(
                summary,
                headers="keys",
                tablefmt="simple_grid",
                showindex=False
            )
        )

        file.write("\n\n")
```
